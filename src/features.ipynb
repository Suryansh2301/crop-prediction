```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
```

```python
def build_preprocessor(kind: str, num_cols, cat_cols):
    """kind='onehot' for linear/KNN models, 'ordinal' for tree models.
    Ordinal output column order is num_cols + cat_cols (used by the explainability code)."""
    if kind == "onehot":
        return ColumnTransformer([
            ("num", StandardScaler(), list(num_cols)),
            ("cat", OneHotEncoder(handle_unknown="ignore"), list(cat_cols)),
        ])
    return ColumnTransformer([
        ("num", "passthrough", list(num_cols)),
        ("cat", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1), list(cat_cols)),
    ])
```
